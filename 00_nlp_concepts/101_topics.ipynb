```python
#### One hot encoding
- size of vector same as vocubulary size
- spars matrix and overfitting issue
- not used in practice
- out of vocubulary issue

#### Bag of words
- uni gram is same as bag or words
- size of vector same as vocubulary size
- spars matrix issue
- out of vocubulary issue is resolved by ignoring any oov word
- semantic  meaning partially added

#### n gram
- size of vector same as vocubulary size
- spars matrix issue
- out of vocubulary issue is resolved by ignoring any oov word
- semantic meaning partially added

#### TF-IDF
- term frequency is between 0 to 1, probability of word in each sentance
- inverse document frequency, log is taken for scaling
- sparse matrix issue
- seems better than previoius methods
- semantic meaning partially added


#### Custom feature
- domain based
- word in text
- hello in text
- average size of text
- number of vovels
- number of consonents
```

```python
#### Word2Vec
- Prediction based
- Deep learning neural network based
- low dimention vector
- non sparsh vector
- no overfitting issue
- semantic meaning captured
- dense embedding

- CBOW
- skip gram

```
